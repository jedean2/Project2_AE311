```python
#Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Lift slope function
def lift_slope(a_0, AR, tau):
    a = a_0 / (1 + (a_0/(np.pi*AR))*(1+tau))
    return a    
```

$C_l$ vs $\alpha$ plots for all three airfoil types
$$C_l  = a_0(\alpha-\alpha_{L0})$$

```python
##rectangular airfoil
a_L0 = -0.036
AR = 4.6
C_la = lift_slope(2*np.pi,AR,0.17)
print(C_la)
a_deg = np.linspace(np.rad2deg(a_L0),10,100)
a_rad = np.deg2rad(a_deg)

#equation
c_l = C_la * (a_rad-a_L0)

##Tapered wing
C_lat = lift_slope(2*np.pi,AR,0.04)
print(C_lat)

#equation tapered
c_l_t = C_lat * (a_rad-a_L0)

##Elliptical 
C_la_e = lift_slope(2*np.pi,AR,0)

#equation elliptical
c_l_e = C_la_e * (a_rad-a_L0)

#plot
plt.plot(a_deg,c_l)
plt.plot(a_deg,c_l_t)
plt.plot(a_deg,c_l_e)   
plt.xlabel('Angle of Attack (degrees)')
plt.ylabel('Lift Coefficient')
plt.title('Lift Curves')
plt.legend(['Rectangular','Tapered','Elliptical'])
plt.grid()
plt.show()


```
